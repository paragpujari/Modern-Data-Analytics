##### Operators

1.  An Operator can be defined as some kind of symbols that is used to perform certain operations on the data.

2.  It is basically used to perform mathematical opeartions on the nuemrical data.


##### Main Operators Categories

| Category   | Operators                  | Purpose                   |
| ---------- | -------------------------- | ------------------------- |
| Arithmetic | `+ - * / // % **`          | Mathematical calculations |
| Comparison | `== != > < >= <=`          | Compare values            |
| Assignment | `= += -= *= /= //= %= **=` | Assign/update values      |
| Logical    | `and or not`               | Combine conditions        |
| Bitwise    | `& \| ^ ~ << >>`           | Work at bit level         |
| Membership | `in`, `not in`             | Check membership          |
| Identity   | `is`, `is not`             | Check object identity     |


##### Types of Operators

1.  Arithmetic Operators ------->  It is the kind of operator that is used to perform certain mathematical or arithmetic operations on the data.

```python
a = 10
b = 3

##### perform arithmetic operations
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)  ##### removes all the values after decimals
print(a%b)   ##### gets the remainder
print(a**b)  ##### exponent of a number
```

2.  Comparison operators ----->  It is used to perform certain kind of comparisons with two or more operands in the data.

```python
a = 10
b = 3

##### perform comparison operations
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)
```

3.  Assignment Operators -------> It is used to perform all kinds of arithmetic operations using assignment.

```python
a = 10
a+=3

print(f'Value of a is {a}')

b = 10
b-=3

print(f'Value of b is {b}')

c = 10
c *= 3

print(f'Value of c is {c}')

d = 10
d /= 3

print(f'Value of d is {d}')

##### Removing all the decimals
e = 10
e //= 3

print(f'Value of e is {e}')

##### Get the remainder
f = 10
f %= 3
print(f'Value of f is {f}')

##### Get the exponent
g = 10
g **=3
print(f'Value of g is {g}')
```

4.  Logical Operators ------->  It is used to perform logical operations with one or more operands.

```python
print((10 > 5) and (8 > 6))

print((2 > 5) or (8 > 10))

print((2>1) and not(10<9))
```

5.  Bitwise Operators ----------->  It is used to perform bitwise operations with one or more operands in the data.

```python
a = 5
b = 3

##### perform bitwise AND
print(a&b)

##### perform bitwise OR
print(a|b)

##### perform bitwise XOR
print(a^b)

##### perform bitwise NOT
print(~(-13))

##### perform bitwise left shift
print(a<<1)

##### perform bitwise right shift
print(a>>1)
```

6.  Membership operators ------> It is a special kind of operator that is used to check whether the entity is a member of the collection of elements or not.

    (a.)    in --------> It checks if the entity is a member of the collection of elements

    (b.)    not in ----> It check if the entity is not a member of the collection of elements.

```python
print(3 in [1,2,3])   

print(4 in [1,2])

print(5 not in [1,2,3,4])

print(10 not in [1,2,3])
```

7.  identity operators ----> It is used to check if both the entities have the same memory address or not.

(a.)   is ----->  It check if both the entities have the same memory address

(b.)   is not ---> It check both the entities does not  have the same memory address

```python
x = [1,2,3]

y = x

print(f'Memory address of x is {id(x)}')
print(f'Memory address of y is {id(y)}')

print(x is y)

print('------------------------------------is not---------------------------------------------')

x = [1,2,3,4,5]
y = [1,2,3,4,5]

print(f'Memory address of x is {id(x)}')
print(f'Memory address of y is {id(y)}')

print(x is not y)
```

##### OBSERVATIONS:

1.  The first instance when y = x , both x and y are pointing to the same memory address.

2.  The second instance when x and y are both having the same collection of elements in the list, then both of its memory address are not the same.

Problem

Suppose:

READ = 1

WRITE = 2

A user has permission value 3.

Check whether the user has the WRITE permission using a bitwise operator

```python
READ = 1

WRITE = 2

permission = 3

##### Check whether the user has the WRITE permission using a bitwise operator
res = (WRITE & permission)

print(f'Value of the result is {res}')
```
