21. Find the first character that appears twice

```python
x = 'programming'
freq={}
##### To store the frequency of each and every character in the dictionary
for i in x:
    if(i in freq):
        freq[i] =   freq[i] + 1
    else:
        freq[i] = 1
print(freq)

##### Find the first character that appears twice
for x,y in freq.items():
    if(y>1):
        print(f'The first character that appears twice is {x}')
        break
```

39.  Compress consecutive characters

```python
x = 'aaabbccc'
x = x.lower()
print(x)
c = 1
res=''
##### To compress the consecutive characters
for i in range(1,len(x)):
    if(x[i]==x[i-1]):
        c = c + 1
    else:
        res = res+x[i-1]+str(c)
        c = 1
res = res+x[i-1]+str(c)
print(res)
```

40. Find the longest palindromic substring

```python
s = "babad"
longest = ""

for i in range(len(s)):
    for j in range(i + 1, len(s) + 1):
        sub = s[i:j]

        if sub == sub[::-1]:
            if len(sub) > len(longest):
                longest = sub

print(longest)
```
