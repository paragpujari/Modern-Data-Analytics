##### Type Conversion

It refers to the technique of converting the data from one data type to another.

##### Loss of precision

It refers to the condition in which there is a loss of data when we convert from higher to lower data type but no loss of precision when we convert the data from lower to higher data type

Ex:---    Float ---------> Int    (loss of precision)

          Int   ---------> Float  (No loss of precision)

```python
##### Perform a type conversion from Int ----> Float
x = 10
print(f'Original type is {type(x)}')
ans = float(x)   ##### int to float
print(ans)
print(f'Converted type is {type(ans)}')
```

##### OBSERVATIONS

1.  As we are converting the data from integer to float(lower to higher data type), so there is no loss of precision.

```python
##### Perform a type conversion from float -----> int
x = 10.23
print(x)
print(f'Original type is {type(x)}')
ans = int(x)    ##### float to int
print(ans)
print(f'Converted type is {type(ans)}')
```

##### OBSERVATIONS:

1.  As we are converting the data from float to integer(higher to lower data type), there is a loss of precision.

##### Types of Type Conversion

1.  Implicit Type Conversion

It is a kind of type conversion where the data of one type is automatically converted into another type.

```python
x = 10
print(f'Original type is {type(x)}')
y = 0.75
#### convert the int type of x into float type in result
ans = x+y
print(ans)
print(f'Converted type is {type(ans)}')
```

##### OBSERVATIONS:

1.  Initially we have x as 10 which is integer. Then we are adding 0.75 to it and then storing the result in res.

2.  Then the res value is in float type.

2.  Explicit Type conversion

Here the programmer needs to manually convert the data of one type into another type.

```python
x = "10.75"  ##### float in string type
print(x)
print(f'Original type is {type(x)}')

ans = float(x)
print(ans)
print(f'Converted type is {type(ans)}')
```

```python
x = 1
print(x)
print(f'Original type is {type(x)}')

ans = bool(x)
print(ans)
print(f'Converted type is {type(ans)}')
```

37.  Perform a complete type-conversion chain

string → integer → float → string

```python
##### Consider a string
x = '10'
print(x)
print(f'Original type of x is {type(x)}')

##### convert string -> integer
ans = int(x)
print(ans)
print(f'Converted type of result is {type(ans)}')

##### convert integer -> float
ans = float(ans)
print(ans)
print(f'Converted type of result is {type(ans)}')

##### convert float -> string
ans = str(ans)
print(ans)
print(f'Converted type of result is {type(ans)}')
```
