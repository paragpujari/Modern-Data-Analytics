##### Rules for defining the Variable

1.  A Variable should always begin with letter or underscore.

2.  It should not begin with the digit.

3.  It may have underscore between the names.

4.  It should not be the name of any keyword.

5.  It should not have hyphen in between the names of the variables.

6.  It should not have any special characters like @,!,#,$,% etc.

```python
##### Rule 1: A variable must begin with an alphabet or an underscore

name = 'Parag'        ##### begins with alphabet
_name = 'Parag'       ##### begins with underscore

print(f'Begins with alphabet is {name}')
print(f'Begins with underscore is {_name}')
```

```python
##### Rule 2: It should not begin with a digit

#1name = 'Parag'   ##### SyntaxError: invalid decimal literal 
```

```python
##### Rule 3: It must have an underscore in between if needed
first_name = 'Parag'
last_name  = 'Pujari'

full_name = first_name + ' ' + last_name
print(f'My full name is {full_name}')
```

```python
##### Rule 4: It should not be any keyword

# def = 'Parag'  ##### SyntaxError: invalid syntax
```

```python
##### Rule 5: It should not have hyphen in between
#first-name = 'Parag'   ##### SyntaxError: cannot assign to expression here. Maybe you meant '==' instead of '='?
```

```python
##### Rule 6: It should not used any special character like ($,%,&) etc
#$name = 'Parag'
#%name = 'Parag'
#&name = 'Parag'     ##### SyntaxError: invalid syntax
```

40. You have the following variables:

original_price = 1000

discount = 200

tax = 80

Calculate the final price using only variables and arithmetic.

The logic is:

Price after discount = original price - discount

Final price = price after discount + tax

```python
original_price = 1000

discount = 200

tax = 80

##### Calculate the price after discount
price_after_discount = original_price - discount
print(f'Price after discount is {price_after_discount}')


##### Calculate the final price
final_price = price_after_discount + tax
print(f'Final price is {final_price}')
```
