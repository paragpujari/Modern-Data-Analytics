##### List Comprehensions

1.  A List comprehension refers to the shorter way of creating the list from the existing iterable.

##### Syntax

answer = [expression for item in iterable if condition]

```python
##### Given a list, find out the square of all the elements in the list
lists = [1,2,3,4,5]

##### Get the square of all the elements in the list
ans = [(i**2) for i in lists if(i != 0)]

print(ans)
```
